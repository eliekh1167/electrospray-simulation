```python
import numpy as np
import matplotlib.pyplot as plt

eps0 = 8.854e-12                    # vacuum permittivity, F/m
theta = np.radians(49.3)            # Taylor cone half-angle
cos_theta = np.cos(theta)
```

```python
def onset_voltage(gamma, r_c, d):
    return np.sqrt((r_c * gamma * cos_theta) / (2 * eps0)) * np.log(4 * d / r_c)
```

```python
propellants = {
    'EMI-BF4':   {'gamma': 0.0452,  'K': 1.4},
    'EMI-GaCl4': {'gamma': 0.0486,  'K': 2.2},
    'EMI-Tf2N':  {'gamma': 0.0349,  'K': 0.88},
    'EMI-BETI':  {'gamma': 0.02875, 'K': 0.34}
}
```

```python
r_c = 10e-6   # tip radius, meters (10 micrometers)
d = 2e-3      # electrode gap, meters (2 mm)

for name, props in propellants.items():
    V = onset_voltage(props['gamma'], r_c, d)
    print(f"{name}: {V/1000:.2f} kV")
```

```python
radii = np.linspace(2e-6, 30e-6, 20)

plt.figure(figsize=(6,4))
for name, props in propellants.items():
    volts = [onset_voltage(props['gamma'], r, d) / 1000 for r in radii]
    plt.plot(radii * 1e6, volts, label=name)

plt.xlabel('Tip radius (µm)')
plt.ylabel('Onset voltage (kV)')
plt.legend()
plt.title('Onset voltage sensitivity to tip radius')
plt.show()
```

```python
from ipywidgets import interact, FloatSlider

@interact(
    rc_um=FloatSlider(min=2, max=30, step=1, value=10, description='Tip radius (µm)'),
    d_mm=FloatSlider(min=0.5, max=5, step=0.1, value=2, description='Gap (mm)')
)
def plot_live(rc_um=10, d_mm=2):
    rc = rc_um * 1e-6
    d = d_mm * 1e-3
    volts = [onset_voltage(propellants[n]['gamma'], rc, d) / 1000 for n in propellants]
    plt.figure(figsize=(6,4))
    plt.bar(propellants.keys(), volts, color=['#2a78d6', '#1baf7a', '#eb6834', '#eda100'])
    plt.ylabel('Onset voltage (kV)')
    plt.ylim(0, 6)
    plt.show()
```
